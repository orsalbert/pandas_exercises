### Step 9. Which was the most-ordered item? 

```python
#groupby item_name, show only quantity column, apply sum
df.groupby('item_name')[['quantity']].sum().sort_values(['quantity'], ascending=False).head(3)
```

```python
#groupby item_name, aggregate on order_id and quantity
df.groupby('item_name').agg(total_orders=('order_id','count'), total_quantity=('quantity', 'sum')).sort_values('total_quantity', ascending = False).head(3)
```

```python
# check what kind of parameters a function has
print(inspect.getfullargspec(df.groupby))
```
