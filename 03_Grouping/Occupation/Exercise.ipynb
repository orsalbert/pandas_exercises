### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user). 

### Step 3. Assign it to a variable called users.

```python
url = 'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user'
df = pd.read_csv(url, sep = '|')
df.head()


```

### Step 4. Discover what is the mean age per occupation

```python
df.groupby('occupation').age.mean().round(1).sort_values(ascending=False)
```

### Step 5. Discover the Male ratio per occupation and sort it from the most to the least

```python
df_gend = df.groupby(['gender','occupation'])[['user_id']].count().unstack(['gender'])
df_mr = pd.DataFrame(round(df_gend.user_id.M / (df.occupation.value_counts().sort_values()) * 100, 3), columns=['male_ratio'])
df_mr.sort_values('male_ratio', ascending=False)
```

### Step 6. For each occupation, calculate the minimum and maximum ages

```python
df.groupby('occupation').age.agg(['min','max'])
```

### Step 7. For each combination of occupation and gender, calculate the mean age

```python
df.groupby(['occupation','gender']).age.mean().unstack().round(2)
```

### Step 8.  For each occupation present the percentage of women and men

```python
df_gend['male_ratio'] = round(df_gend.user_id.M/df.occupation.value_counts() * 100, 2)
df_gend['female_ratio'] = round(df_gend.user_id.F/df.occupation.value_counts() * 100, 2)

df_gend
```
