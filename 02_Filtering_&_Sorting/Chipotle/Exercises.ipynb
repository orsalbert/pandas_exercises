### Step 3. Assign it to a variable called chipo.

```python
url = 'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv'
df = pd.read_csv(url, sep='\t')
```

### Step 4. How many products cost more than $10.00?

```python
df['item_price'] = df.item_price.apply(lambda x: float(x[1:]))
df['price_per_item'] = df.item_price / df.quantity
df.query('price_per_item > 10').item_name.nunique()
```

### Step 5. What is the price of each item? 
###### print a data frame with only two columns item_name and item_price

```python
df[['item_name', 'price_per_item']].drop_duplicates().sort_values(['price_per_item'])
```

### Step 6. Sort by the name of the item

```python
df.sort_values(['item_name'])
```

### Step 7. What was the quantity of the most expensive item ordered?

```python
# df.sort_values('price_per_item', ascending=False)
# df[df['price_per_item'] == df.price_per_item.max()]['quantity'].sum()
# df[df['price_per_item'] == df.price_per_item.max()].agg(sum=('quantity','sum'))

df[df['item_price'] == df.item_price.max()]
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
df[df['item_name'] == 'Veggie Salad Bowl'].quantity.sum()
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
df[(df['item_name'] == 'Canned Soda') & (df['quantity'] > 1)].order_id.count()
```
