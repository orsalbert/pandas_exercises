### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# set this so the graphs open internally
%matplotlib inline
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

### Step 3. Assign it to a variable called chipo.

```python
df = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv', sep = '\t')
df.head()
```

### Step 4. See the first 10 entries

```python
df.head()
```

### Step 5. Create a histogram of the top 5 items bought

```python
# Counter(df.item_name)

top_five = df.groupby('item_name')['item_name'].count().sort_values(ascending=False).head(5)
top_five.plot(kind='bar')

plt.xlabel('Items')
plt.ylabel('Number of items ordered')
plt.title('Most ordered items')
```

### Step 6. Create a scatterplot with the number of items orderered per order price
#### Hint: Price should be in the X-axis and Items ordered in the Y-axis

```python
# df.item_price = [float(i[1:]) for i in df.item_price]

df_orders = df.groupby('order_id').sum()

plt.scatter(x=df_orders.item_price, y=df_orders.quantity, s = 50, c = 'green')

plt.xlabel('Order Price')
plt.ylabel('Items ordered')
plt.title('Number of items ordered per order price')
plt.ylim(0)
```
